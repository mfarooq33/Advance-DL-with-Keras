# **Text Classification**


```python
from sklearn.datasets import fetch_20newsgroups
import pandas as pd

# Load dataset
newsgroups = fetch_20newsgroups(subset='all', categories=['rec.sport.baseball', 'sci.space'], shuffle=True, random_state=42)
data = newsgroups.data
target = newsgroups.target

# Create a DataFrame for easy manipulation
df = pd.DataFrame({'text': data, 'label': target})
df.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TF-IDF Vectorizer
vectorizer = TfidfVectorizer(stop_words='english', max_df=0.7)

# Transform the text data to feature vectors
X = vectorizer.fit_transform(df['text'])

# Labels
y = df['label']
print(y)
```

```python
print(X.shape)
```

```python
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize and train the classifier
clf = SVC(kernel='linear')
clf.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Predict on the test set
y_pred = clf.predict(X_test)

# Evaluate the performance
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=newsgroups.target_names)

print(f'Accuracy: {accuracy:.4f}')
print('Classification Report:')
print(report)
```

```python
def predict_category(text):
    """
    Predict the category of a given text using the trained classifier.
    """
    text_vec = vectorizer.transform([text])
    prediction = clf.predict(text_vec)
    return newsgroups.target_names[prediction[0]]

# Example usage
sample_text = "NASA announced the discovery of new exoplanets."
predicted_category = predict_category(sample_text)
print(f'The predicted category is: {predicted_category}')
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Get precision, recall, and f1-score from the classification report
report_data = classification_report(y_test, y_pred, target_names=newsgroups.target_names, output_dict=True)
metrics = ['precision', 'recall', 'f1-score']
categories = newsgroups.target_names
bar_width = 0.25
x = np.arange(len(categories))

fig, ax = plt.subplots(figsize=(10, 6))

for i, metric in enumerate(metrics):
    values = [report_data[cat][metric] for cat in categories]
    ax.bar(x + i * bar_width, values, bar_width, label=metric)

# Add labels, title, and customize the plot
ax.set_ylabel('Score')
ax.set_title('Classification Report Metrics by Category')
ax.set_xticks(x + bar_width)
ax.set_xticklabels(categories)
ax.legend()
ax.set_ylim(0, 1.1) # Set y-axis limit from 0 to 1.1 to better visualize scores

plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Calculate the confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Display the confusion matrix
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=newsgroups.target_names)
disp.plot()
plt.title('Confusion Matrix')
plt.show()
```

```python
import pandas as pd
import torch
import numpy as np
from scipy.special import expit
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from collections import Counter
import re
import warnings
warnings.filterwarnings('ignore')

# Sample dataset of tweets
def create_sample_tweets():
    """Create a sample dataset of tweets for demonstration"""
    tweets = [
        "Just bought the new iPhone and it's amazing! The camera quality is outstanding. #Apple #Tech",
        "When will this pandemic end? I'm so tired of all these restrictions and lockdowns.",
        "Can anyone recommend a good Italian restaurant in downtown? Looking for dinner options.",
        "The government should really focus more on climate change policies. Our future is at stake!",
        "Just finished my morning run. Feeling great and energized for the day! #fitness #health",
        "Why is the customer service so bad these days? Been on hold for 45 minutes!",
        "Looking for recommendations for a good fantasy book series. Any suggestions?",
        "Stock market is crashing again. Should I sell my investments or wait it out?",
        "Just watched the latest Marvel movie. The special effects were incredible!",
        "Dealing with technical issues on my laptop. Can anyone help with Windows troubleshooting?",
        "The new tax reforms are going to hurt middle-class families. This is unacceptable!",
        "Beautiful sunset at the beach today! #nature #photography",
        "How do I apply for a mortgage with bad credit? Any advice would be helpful.",
        "Another school shooting in the news. When will we get proper gun control laws?",
        "My favorite band just announced a world tour! Can't wait to get tickets. #music"
    ]

    df = pd.DataFrame({'text': tweets})
    return df

# Improved intent classification
def get_intent(text):
    """
    Enhanced intent classification with better pattern matching
    """
    text_lower = text.lower()

    # Question patterns
    question_patterns = [
        r'\b(can anyone|can somebody|any suggestions|any advice|how do|how to|what is|where can|when will|why is)\b',
        r'\?'
    ]

    # Complaint patterns
    complaint_patterns = [
        r'\b(bad|terrible|awful|horrible|annoying|frustrating|problem|issue|error|broken|wrong)\b',
        r'\b(why is|when will|this is unacceptable)\b'
    ]

    # Recommendation patterns
    recommendation_patterns = [
        r'\b(recommend|suggestion|advice|looking for)\b'
    ]

    # Check for questions
    for pattern in question_patterns:
        if re.search(pattern, text_lower):
            return "question"

    # Check for complaints
    for pattern in complaint_patterns:
        if re.search(pattern, text_lower):
            return "complaint"

    # Check for recommendations
    for pattern in recommendation_patterns:
        if re.search(pattern, text_lower):
            return "recommendation_request"

    # Default to statement
    return "statement"

# Improved classification function with error handling
def classify_text(text, model_topic, tokenizer_topic, topic_labels):
    """
    Classify the topic and intent of a given text using pre-trained models.

    Args:
        text (str): Input text to classify
        model_topic: Pre-trained topic classification model
        tokenizer_topic: Tokenizer for topic classification
        topic_labels (list): List of topic labels

    Returns:
        tuple: (predicted_topics, intent, confidence_scores)
    """
    if not text or pd.isna(text):
        return [], "unknown", []

    try:
        # Topic classification
        inputs = tokenizer_topic(text, return_tensors="pt", truncation=True, max_length=128, padding=True)

        with torch.no_grad():
            outputs = model_topic(**inputs)

        scores = outputs.logits.detach().numpy()
        probs = expit(scores).flatten()

        # Get topics with probability > 0.3 (more lenient threshold)
        predicted_topics = [topic_labels[i] for i, p in enumerate(probs) if p > 0.3]

        # If no topics above threshold, take the top 1
        if not predicted_topics:
            top_idx = np.argmax(probs)
            predicted_topics = [topic_labels[top_idx]]

        # Intent classification
        intent = get_intent(text)

        # Get confidence scores for predicted topics
        confidence_scores = [float(probs[i]) for i, topic in enumerate(topic_labels) if topic in predicted_topics]

        return predicted_topics, intent, confidence_scores

    except Exception as e:
        print(f"Error classifying text: {e}")
        return [], "error", []

# Main execution function
def main():
    # Create sample dataset
    print("Creating sample dataset of tweets...")
    df_tweets = create_sample_tweets()

    # Display original dataset
    print("Sample Tweets Dataset:")
    print(df_tweets)
    print(" + "="*80 + ")

    # For demonstration purposes, we'll create mock models and tokenizers
    # In real usage, you would load actual pre-trained models

    class MockModel:
        def __init__(self, num_labels):
            self.num_labels = num_labels

        def __call__(self, **kwargs):
            class MockOutput:
                def __init__(self, num_labels):
                    # Generate random logits
                    self.logits = torch.randn(1, num_labels)
            return MockOutput(self.num_labels)

    class MockTokenizer:
        def __call__(self, text, return_tensors, truncation, max_length, padding):
            return {'input_ids': torch.randint(0, 1000, (1, 10)),
                   'attention_mask': torch.ones(1, 10)}

    # Mock topic labels (common topics in tweets)
    topic_labels = [
        "technology", "health", "politics", "entertainment",
        "sports", "business", "food", "travel", "education", "other"
    ]

    # Initialize mock models (replace with actual models in production)
    model_topic = MockModel(len(topic_labels))
    tokenizer_topic = MockTokenizer()

    print("Classifying tweets...")

    # Apply classification with progress tracking
    results = []
    for idx, text in enumerate(df_tweets['text']):
        topics, intent, confidence = classify_text(
            text, model_topic, tokenizer_topic, topic_labels
        )
        results.append({
            'text': text,
            'predicted_topics': topics,
            'predicted_intent': intent,
            'confidence_scores': confidence
        })

        # Print progress for sample dataset
        print(f"Processed {idx + 1}/{len(df_tweets)}: {intent} - {topics}")

    # Create results DataFrame
    df_results = pd.DataFrame(results)

    # Display results
    print(" + "="*80)
    print("CLASSIFICATION RESULTS:")
    print("="*80)

    for _, row in df_results.iterrows():
        print(f"\nTweet: {row['text']}")
        print(f"Topics: {row['predicted_topics']}")
        print(f"Intent: {row['predicted_intent']}")
        print(f"Confidence: {[f'{c:.2f}' for c in row['confidence_scores']]}")
        print("-" * 50)

    # Summary statistics
    print(" + "="*80)
    print("SUMMARY STATISTICS:")
    print("="*80)

    # Intent distribution
    intent_counts = df_results['predicted_intent'].value_counts()
    print(f"\nIntent Distribution:")
    for intent, count in intent_counts.items():
        print(f"  {intent}: {count}")

    # Topic distribution
    all_topics = [topic for sublist in df_results['predicted_topics'] for topic in sublist]
    topic_counts = Counter(all_topics)
    print(f"\nTopic Distribution:")
    for topic, count in topic_counts.most_common():
        print(f"  {topic}: {count}")

    return df_results

# Enhanced version with actual Hugging Face model (commented out for demo)
def classify_with_real_model():
    """
    Example of how to use with real Hugging Face models
    Uncomment and modify based on your specific models
    """
    """
    # Load real models (example with distilbert)
    topic_model_name = "distilbert-base-uncased"  # Replace with your actual model
    tokenizer_topic = AutoTokenizer.from_pretrained(topic_model_name)
    model_topic = AutoModelForSequenceClassification.from_pretrained(topic_model_name)

    # Your actual topic labels
    topic_labels = ["label1", "label2", "label3"]  # Replace with your actual labels

    # Classify with real model
    df_tweets = create_sample_tweets()
    results = []
    for text in df_tweets['text']:
        topics, intent, confidence = classify_text(text, model_topic, tokenizer_topic, topic_labels)
        results.append({
            'text': text,
            'predicted_topics': topics,
            'predicted_intent': intent,
            'confidence_scores': confidence
        })

    return pd.DataFrame(results)
    """

if __name__ == "__main__":
    #Run the main classification
    df_classified = main()

      #Uncomment below to use with real models
    #df_classified_real = classify_with_real_model()
```

```python
import nltk
from nltk.corpus import wordnet
import random

nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt')
nltk.download('punkt_tab')

def get_synonyms(word):
    """Get synonyms for a word from WordNet."""
    synonyms = set()
    for syn in wordnet.synsets(word):
        for lemma in syn.lemmas():
            synonym = lemma.name().replace('_', ' ').lower()
            if synonym != word:
                synonyms.add(synonym)
    return list(synonyms)

def synonym_replacement(sentence, n=1):
    """Replace n random non-stopwords with their synonyms."""
    words = nltk.word_tokenize(sentence)
    new_words = words.copy()
    random_word_list = list(set([word for word in words if wordnet.synsets(word)]))
    random.shuffle(random_word_list)
    num_replaced = 0

    for random_word in random_word_list:
        synonyms = get_synonyms(random_word)
        if len(synonyms) >= 1:
            synonym = random.choice(synonyms)
            new_words = [synonym if word == random_word else word for word in new_words]
            num_replaced += 1
        if num_replaced >= n:
            break

    return ' '.join(new_words)

# Original Tweet
original_tweet = "I really enjoyed the concert last night. The band was amazing!"
# Augmented Tweet
augmented_tweet = synonym_replacement(original_tweet, n=2)
print(f"Original: {original_tweet}")
print(f"Augmented: {augmented_tweet}")
# Example Output: "I truly enjoyed the concert last night. The band was fantastic!"
```

```python
# STA Module Implementation for Few/Zero-Shot Learning with Pseudo-Labeling,
## Overview,
#This notebook implements the **Spatio-Temporal Aggregation (STA) Module** from the paper *Spatio-Temporal Self-supervision for Few-Shot Action Recognition* (2023). The STA module uses self-attention to aggregate spatial relationships across video frames for robust embeddings in few-shot settings.,
#We integrate it into a **pseudo-labeling pipeline** for few/zero-shot learning:,
#Few-Shot**: Train on small labeled video data.,
#Zero-Shot**: Skip initial training (use random/pre-trained weights).,
#Pseudo-Labeling**: Generate & filter high-confidence labels for unlabeled data, then retrain iteratively.,

 #vData**: Synthetic spatio-temporal sequences (T=8 frames, H=W=32, C=3 channels) simulating actions. Real data: Swap with UCF101 via `torchvision`.,

 #Requirements**: PyTorch 1.12+, NumPy, scikit-learn."

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import numpy as np
from sklearn.model_selection import train_test_split
from torch.utils.data import DataLoader, TensorDataset

# Set seeds for reproducibility
torch.manual_seed(42)
np.random.seed(42)

def generate_synthetic_video_data(n_labeled_per_class=5, n_unlabeled=200, n_test_per_class=20, seq_len=8, height=32, width=32, channels=3):
    # Class 0: No-action (static noise)
    X_class0 = np.random.normal(0, 1, (n_labeled_per_class + n_test_per_class + n_unlabeled//2, channels, seq_len, height, width))
    y_class0 = np.zeros(n_labeled_per_class + n_test_per_class + n_unlabeled//2)

    # Class 1: Action (moving sine wave pattern across frames)
    X_class1 = np.zeros((n_labeled_per_class + n_test_per_class + n_unlabeled//2, channels, seq_len, height, width))
    for t in range(seq_len):
        x = np.linspace(-np.pi, np.pi, width)
        y = np.linspace(-np.pi, np.pi, height)
        X_class1[:, 0, t, :, :] = np.sin(x + t * 0.5)[:, np.newaxis] * np.cos(y[np.newaxis, :] + t * 0.3)
        X_class1[:, 1:, t, :, :] = np.random.normal(0, 0.5, (n_labeled_per_class + n_test_per_class + n_unlabeled//2, 2, height, width))
    y_class1 = np.ones(n_labeled_per_class + n_test_per_class + n_unlabeled//2)

    # Combine classes
    X = np.concatenate([X_class0, X_class1], axis=0)
    y = np.concatenate([y_class0, y_class1], axis=0)

    # Normalize to [0,1]
    X = (X - X.min()) / (X.max() - X.min() + 1e-8)

    # Split indices
    class0_idx = np.where(y == 0)[0]
    class1_idx = np.where(y == 1)[0]

    # Labeled (few-shot)
    labeled0_idx = np.random.choice(class0_idx, n_labeled_per_class, replace=False)
    labeled1_idx = np.random.choice(class1_idx, n_labeled_per_class, replace=False)
    labeled_idx = np.concatenate([labeled0_idx, labeled1_idx])
    X_labeled = X[labeled_idx]
    y_labeled = y[labeled_idx]

    # Test
    test0_idx = np.random.choice(np.setdiff1d(class0_idx, labeled0_idx), n_test_per_class, replace=False)
    test1_idx = np.random.choice(np.setdiff1d(class1_idx, labeled1_idx), n_test_per_class, replace=False)
    test_idx = np.concatenate([test0_idx, test1_idx])
    X_test = X[test_idx]
    y_test = y[test_idx]

    # Unlabeled (remaining)
    unlabeled_idx = np.random.choice(np.setdiff1d(np.arange(len(X)), np.concatenate([labeled_idx, test_idx])), n_unlabeled, replace=False)
    X_unlabeled = X[unlabeled_idx]
    y_unlabeled_true = y[unlabeled_idx]  # For eval only

    # To tensors
    X_labeled = torch.FloatTensor(X_labeled)
    y_labeled = torch.LongTensor(y_labeled)
    X_unlabeled = torch.FloatTensor(X_unlabeled)
    y_unlabeled_true = torch.LongTensor(y_unlabeled_true)
    X_test = torch.FloatTensor(X_test)
    y_test = torch.LongTensor(y_test)

    print(f"Labeled: {X_labeled.shape[0]} samples, Unlabeled: {X_unlabeled.shape[0]}, Test: {X_test.shape[0]}")
    return X_labeled, y_labeled, X_unlabeled, y_unlabeled_true, X_test, y_test

# Generate data
X_labeled, y_labeled, X_unlabeled, y_unlabeled_true, X_test, y_test = generate_synthetic_video_data()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score
from torch.utils.data import DataLoader, TensorDataset
import urllib.request
import zipfile
import os

# Set seeds for reproducibility
torch.manual_seed(42)
np.random.seed(42)

def download_crisislex_t6(save_path='./crisislex_t6.zip'):
    if not os.path.exists(save_path):
        url = 'https://github.com/sajao/CrisisLex/raw/master/releases/CrisisLexT6-v1.0.zip'
        print(f'Downloading from {url}...')
        try:
            urllib.request.urlretrieve(url, save_path)
            print('Downloaded.')
        except Exception as e:
            print(f'Download failed: {e}')
            return False
    else:
        print('Dataset exists.')
    return True

def load_flood_dataset(data_dir='./data', n_labeled=50, n_unlabeled=1000):
    os.makedirs(data_dir, exist_ok=True)
    zip_path = os.path.join(data_dir, 'CrisisLexT6-v1.0.zip')
    download_success = download_crisislex_t6(zip_path)

    if download_success:
        extract_dir = os.path.join(data_dir, 'CrisisLexT6')
        if not os.path.exists(extract_dir):
            with zipfile.ZipFile(zip_path, 'r') as z:
                z.extractall(data_dir)
        csv_path = os.path.join(extract_dir, '2013_AlbertaFloods_keywordSampled.csv')
        try:
            df = pd.read_csv(csv_path, encoding='latin-1', names=['TweetID', 'TweetDate', 'TweetText', 'Class', 'InfoType', 'Source'])
            df = df.dropna(subset=['TweetText', 'Class'])
            df['text'] = df['TweetText'].astype(str)
            df['label'] = (df['Class'] == 'on-topic').astype(int)
            print(f'Loaded {len(df)} tweets. Labels: {df["label"].value_counts().to_dict()}')
        except FileNotFoundError:
            print(f'CSV not found at {csv_path}. Using synthetic data.')
            download_success = False

    if not download_success:
        # Synthetic fallback
        np.random.seed(42)
        on_topic = ['Flood in Alberta is devastating #AlbertaFloods', 'Evacuations due to rising waters', 'Help needed in flooded areas'] * 1500
        off_topic = ['Watching TV tonight', 'Nice weather today', 'Cooking dinner'] * 2000
        texts = on_topic + off_topic
        labels = [1] * len(on_topic) + [0] * len(off_topic)
        df = pd.DataFrame({'text': texts, 'label': labels})
        print(f'Synthetic: {len(df)} tweets. Labels: {df["label"].value_counts().to_dict()}')

    # Splits
    on_idx = df[df['label'] == 1].index
    off_idx = df[df['label'] == 0].index
    labeled_on = np.random.choice(on_idx, min(n_labeled//2, len(on_idx)), replace=False)
    labeled_off = np.random.choice(off_idx, min(n_labeled//2, len(off_idx)), replace=False)
    labeled_idx = np.concatenate([labeled_on, labeled_off])

    rem_on = np.setdiff1d(on_idx, labeled_on)
    rem_off = np.setdiff1d(off_idx, labeled_off)
    test_on = np.random.choice(rem_on, min(100, len(rem_on)), replace=False)
    test_off = np.random.choice(rem_off, min(100, len(rem_off)), replace=False)
    test_idx = np.concatenate([test_on, test_off])

    rem_idx = np.setdiff1d(np.arange(len(df)), np.concatenate([labeled_idx, test_idx]))
    unlab_idx = np.random.choice(rem_idx, min(n_unlabeled, len(rem_idx)), replace=False)

    X_labeled = df.loc[labeled_idx, 'text'].values
    y_labeled = df.loc[labeled_idx, 'label'].values
    X_unlabeled = df.loc[unlab_idx, 'text'].values
    y_unlabeled_true = df.loc[unlab_idx, 'label'].values
    X_test = df.loc[test_idx, 'text'].values
    y_test = df.loc[test_idx, 'label'].values

    X_train, X_val, y_train, y_val = train_test_split(X_labeled, y_labeled, test_size=0.2, stratify=y_labeled)

    return X_train, X_val, y_train, y_val, X_unlabeled, y_unlabeled_true, X_test, y_test # Explicitly return variables
```
